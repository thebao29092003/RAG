    Small Chunk Size
        Advantages
            Fine-grained search with higher accuracy for specific queries.
            Useful for datasets with highly distributed information.
        Disadvantages
            Context fragmentation: important context might span multiple chunks.
            Increased resource consumption (storage and compute).

    Large Chunk Size
        Advantages
            Better for context-aware retrieval.
            Reduces embedding computation and storage.
        Disadvantages
            Low recall for specific keyword-based queries.
            Risks of irrelevant content being retrieved.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

# Sample document (larger for better testing)
document = """
Elon Musk is the CEO of Tesla. Tesla's mission is to accelerate the world's transition to sustainable energy.
Tesla manufactures electric vehicles (EVs), battery energy storage, solar panels, and related products. SpaceX, another company led by Musk,
aims to make space travel accessible to humanity by developing reusable rockets.
Musk has also co-founded Neuralink, which focuses on connecting the human brain to computers using advanced neural interfaces.
Recently, Musk has expressed his concerns about artificial intelligence, advocating for regulation to prevent potential harm.
Musk believes that governments should play an active role in ensuring the safe development of AI technologies.
His innovative ventures span automotive, space exploration, energy, AI, and neuroscience, making him a pivotal figure in modern technology.
"""

# chunk_size=10: Mỗi đoạn văn bản (chunk) sau khi cắt sẽ có độ dài tối đa là 10 ký tự.
# chunk_overlap=2: Đoạn sau sẽ lặp lại (gối đầu) 2 ký tự cuối của đoạn trước. Việc này giúp giữ mạch văn, đảm bảo ngữ cảnh không bị cắt đứt đột ngột giữa các đoạn.
# RecursiveCharacterTextSplitter: Đây là bộ cắt thông minh, nó sẽ cố gắng không cắt ngang giữa từ hoặc câu (ưu tiên cắt tại khoảng trắng hoặc dấu dòng) cho đến khi đạt gần mức 10 ký tự.
# split_text(document): Thực hiện hành động cắt nội dung của biến document và lưu kết quả vào danh sách small_chunk.
small_splitter = RecursiveCharacterTextSplitter(chunk_size=10, chunk_overlap=2)
small_chunk = small_splitter.split_text(document)
print("Smaller chunks -")
print(small_chunk)

#large chunk size
large_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=10)
large_chunk = large_splitter.split_text(document)
print("Larger chunks -")
print(large_chunk)

# Initialize embedding model
# embedding_model: Là "bộ thông dịch" biến câu chữ thành các dãy số (Vector). Máy tính không hiểu ý nghĩa của từ, nhưng nó có thể so sánh khoảng cách giữa các dãy số để biết hai câu có ý nghĩa gần nhau hay không.
# all-MiniLM-L6-v2: Một mô hình cực kỳ phổ biến, nhỏ gọn và chạy rất nhanh trên máy tính cá nhân (local). Nó biến mỗi đoạn văn bản thành một vector có 384 chiều.
embedding_model = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Create vector stores
# Lấy dữ liệu: Nhận vào danh sách các đoạn văn bản (small_chunk hoặc large_chunk).
# Tính toán: Dùng embedding_model để tính toán Vector cho từng đoạn văn bản đó.
# Lưu trữ: Lưu cả văn bản gốc và Vector tương ứng của nó vào database Chroma.
small_vector_store = Chroma.from_texts(small_chunk, embedding_model)
large_vector_store = Chroma.from_texts(large_chunk, embedding_model)

# Semantic query
query = "What are Musk's views on artificial intelligence and government regulations?"

# Perform search
small_results = small_vector_store.similarity_search(query, k=1)
large_results = large_vector_store.similarity_search(query, k=1)

# câu trả lời khá giống nhau vì dữ liệu nhỏ, nếu dữ liệu lớn hơn có thể khác

# Display results
print("-----------------------------------")
print("Results")
print("-----------------------------------")
print("Small Chunk Results:", small_results)
print("Large Chunk Results:", large_results)
```

    1. Fixed-size Chunking (Sliding Window)
    What It Is:
    Fixed-size chunking divides the text into fixed-length pieces or windows. This is done by splitting the text into chunks of a predetermined size (e.g., 512 tokens). This strategy is simple but can be inefficient for longer texts, especially if chunks cut across meaningful sections (like sentences).

    Use Case:
    Best for datasets where text length is fairly uniform or where processing uniform chunks is necessary.

```python
#words = text.split():
# Cắt chuỗi văn bản thành một danh sách (list) các từ đơn lẻ.
# Ví dụ: "Hello world" thành ['Hello', 'world']. Mặc định nó sẽ cắt dựa trên khoảng trắng.
# chunks = [words[i:i + chunk_size] for i in range(0, len(words), chunk_size)]:
# Đây là một kỹ thuật "chia mảng thành các phần bằng nhau" (array chunking).
# Nó duyệt qua danh sách từ, mỗi lần nhảy một bước bằng chunk_size.
# Kết quả là một danh sách chứa các danh sách con, mỗi danh sách con có tối đa 5 từ (theo ví dụ).
# return [' '.join(chunk) for chunk in chunks]:
# Gộp các từ trong mỗi danh sách con lại thành một câu hoàn chỉnh bằng dấu cách.
def fixed_size_chunk(text, chunk_size=512):
    # Split text into words and chunk them into fixed-size parts
    words = text.split()
    chunks = [words[i:i + chunk_size] for i in range(0, len(words), chunk_size)]
    print(chunks)
    return [' '.join(chunk) for chunk in chunks]

# Example text (typically a long document)
text = "This is a very long text that should be chunked into smaller pieces for efficient processing."

# Chunk into 5 words per chunk
chunks = fixed_size_chunk(text, chunk_size=8)
print(chunks)

# Display the chunks
for i, chunk in enumerate(chunks):
    print(f"Chunk {i + 1}: {chunk}")
```

    4. Semantic-based Chunking (Dynamic Chunking)
    What It Is:
    Semantic-based chunking involves splitting text into chunks based on the meaning of the content rather than fixed length or sentence boundaries. The idea is to group related information, which may vary in length, into semantic units. This is often done using NLP techniques, such as sentence embeddings or document clustering, to identify the most relevant sections for each chunk.

    Use Case:
    Best for use in situations where meaningful semantic units (e.g., paragraphs, topics) should be grouped together regardless of the chunk size. This is useful for document summarization, knowledge extraction, and other advanced NLP tasks.

    Đây là một ví dụ về kỹ thuật Semantic Chunking (Chia nhỏ dữ liệu theo ngữ nghĩa) bằng cách sử dụng thuật toán phân cụm K-Means.
    Thay vì chia văn bản theo độ dài (ký tự) hay dấu câu, cách này sẽ nhóm các câu có nội dung giống nhau hoặc liên quan đến nhau vào cùng một "chunk" (đoạn).
    Giải thích chi tiết từng phần:
    1. Khởi tạo mô hình Embedding:
        model = SentenceTransformer('all-MiniLM-L6-v2')
        Dùng để biến câu văn thành các Vector (dãy số). Các câu có ý nghĩa giống nhau sẽ có các Vector nằm gần nhau trong không gian toán học.
    2. Tạo Embeddings:
        embeddings = model.encode(sentences)
        Máy tính sẽ tính toán và chuyển 4 câu ví dụ thành 4 dãy số.
    3. Phân cụm bằng KMeans:
        kmeans = KMeans(n_clusters=2, random_state=0)
        labels = kmeans.fit_predict(embeddings)
        n_clusters=2: Bạn yêu cầu thuật toán chia 4 câu này thành 2 nhóm (2 chủ đề).
        fit_predict: Thuật toán sẽ nhìn vào các Vector và tự tính toán xem câu nào nên đi với câu nào dựa trên khoảng cách giữa các Vector đó.
    4. Gom nhóm và Hiển thị:
        Code sử dụng một từ điển (chunks) để gom các câu có cùng nhãn (label) lại với nhau.
        Cuối cùng in ra kết quả của từng nhóm.

```python
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
import numpy as np

# Initialize a model for sentence embeddings
model = SentenceTransformer('all-MiniLM-L6-v2')

# Example text (to simulate chunking)
sentences = [
    "Astronauts are sent to space.",
    "The Martian is about survival on Mars.",
    "Interstellar deals with space exploration.",
    "Space travel involves many challenges."
]

# Create embeddings for each sentence
embeddings = model.encode(sentences)

# Use clustering to group semantically similar sentences
kmeans = KMeans(n_clusters=2, random_state=0)
labels = kmeans.fit_predict(embeddings)

# Group sentences by their cluster label
chunks = {}
for i, label in enumerate(labels):
    if label not in chunks:
        chunks[label] = []
    chunks[label].append(sentences[i])

# Display semantic chunks
for label, chunk in chunks.items():
    print(f"Semantic Chunk {label + 1}: {', '.join(chunk)}")
```

    5. Overlapping Chunking
    What It Is:
    Overlapping chunking involves creating chunks that overlap with each other. For example, one chunk might contain the last few words of the previous chunk. This strategy is used to preserve context between chunks and ensure that no information is lost when dividing long texts.

    Use Case:
    Useful when context between sentences is important, such as for tasks like machine translation, summarization, or question-answering where the meaning of adjacent chunks is highly dependent on each other.

nó cũng giống như chia theo kích thước cố định nhưng mà giờ, câu sau sẽ chứa 1 số từ ở câu trước

```python
def overlapping_chunk(text, chunk_size=5, overlap=2):
    words = text.split()
    chunks = []
    for i in range(0, len(words), chunk_size - overlap):
        chunk = words[i:i + chunk_size]
        chunks.append(' '.join(chunk))
    return chunks

# Example text
text = "This is an example of overlapping chunking to maintain context between chunks."

# Chunk into overlapping pieces with 5 words per chunk and 2-word overlap
chunks = overlapping_chunk(text, chunk_size=5, overlap=2)

# Display the chunks
for i, chunk in enumerate(chunks):
    print(f"Chunk {i + 1}: {chunk}")
```

    6. Recursive Chunking
    Recursive chunking divides text in a hierarchical, iterative manner using a set of separators (such as newline characters or spaces). This method allows for more structured chunking based on content boundaries.

    Explanation:
    The method works by recursively splitting text into smaller chunks by examining separators like \n\n, \n, and spaces. LangChain offers the RecursiveCharacterTextSplitter class, which allows the specification of separators to control how the text is split.


    Đây là bộ chia văn bản (Text Splitter) được khuyến khích sử dụng nhất trong hầu hết các hệ thống RAG vì sự thông minh và linh hoạt của nó.
    Dưới đây là giải thích ngắn gọn về cách hoạt động của đoạn code này:
    1. Tại sao gọi là "Recursive" (Đệ quy)?
    Khác với các bộ chia thông thường chỉ cắt theo một ký tự cố định, bộ chia này hoạt động theo kiểu "Thử và Sai" dựa trên danh sách separators mà bạn cung cấp:
        Đầu tiên, nó cố gắng chia văn bản theo \n\n (đoạn văn).
        Nếu đoạn văn đó vẫn dài hơn chunk_size (50), nó sẽ chia tiếp đoạn đó theo \n (dòng).
        Nếu vẫn còn quá dài, nó lại chia tiếp theo " " (khoảng trắng giữa các từ).
        Cuối cùng, nếu vẫn dài quá, nó mới cắt đến từng ký tự.
    Mục đích: Để giữ các câu và các đoạn có liên quan ở gần nhau nhất có thể, tránh việc cắt ngang giữa chừng một từ hoặc một ý.
    2. Ý nghĩa các thông số
        separators=["\n\n", "\n", " "]: Thứ tự ưu tiên cắt. Nó sẽ ưu tiên cắt ở cấp độ đoạn văn trước, rồi mới đến dòng, rồi mới đến từ.
        chunk_size=50: Độ dài mục tiêu của mỗi đoạn (tính bằng ký tự).
        chunk_overlap=10: Gối đầu 10 ký tự. Đoạn 2 sẽ chứa 10 ký tự cuối của đoạn 1. Điều này giúp AI không bị mất ngữ cảnh ở các điểm cắt.
    3. Phân tích kết quả thực tế từ ví dụ
    Văn bản của bạn có đoạn: "This is a paragraph \n. This is another paragraph..."
        Nó thấy dấu \n\n sau chữ paragraph., nó sẽ cắt tại đó để tạo ra một đoạn lớn.
        Trong đoạn đó, nếu tổng số ký tự chưa đến 50, nó sẽ giữ nguyên. Nếu vượt quá 50, nó sẽ tìm các dấu \n hoặc khoảng trắng bên trong để chia nhỏ thêm.
    Ưu điểm vượt trội:
        Giữ ngữ cảnh tốt: Nó cố gắng tối đa để không làm nát câu văn của bạn.
        Linh hoạt: Phù hợp với nhiều loại văn bản từ bài báo, sách đến mã nguồn.

    Lời khuyên: Trong các dự án RAG thực tế, người ta thường để chunk_size khoảng 500 - 1000 và chunk_overlap khoảng 10% - 20% kích thước chunk để đạt hiệu quả tìm kiếm tốt nhất. Con số 50 trong ví dụ của bạn là rất nhỏ, thường chỉ dùng để test.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Sample text
text = """This is a paragraph \n. This is another paragraph. This is a new paragraph.

Here is some additional content.\n"""

# Initialize the RecursiveCharacterTextSplitter with separators
splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", " "],
    chunk_size=50,  # Adjust chunk size as necessary
    chunk_overlap=10  # Optional: overlap chunks for context preservation
)

# Split the text
chunks = splitter.split_text(text)

# Display chunks
for i, chunk in enumerate(chunks):
    print(f"Chunk {i + 1}: {chunk}")
```

    7. Agentic Chunking
    Agentic chunking leverages large language models (LLMs) to determine how to chunk the text based on its context. This is a more dynamic approach where the model itself decides how much and what part of the text should form a chunk.

    Explanation:
    The LLM is tasked with understanding the context of the text and then creating chunks based on the relevance of the information. This technique uses the model's understanding of the content to ensure that chunks are informative and relevant.


    Đây là phương pháp agentic Chunking (Chia nhỏ dựa vào LLM) để quyết định chia như thế nào
    1. Giải thích các thành phần chính:
        tokenizer (t5-base): Là bộ mã hóa biến văn bản thành các con số (Token IDs). T5 là một mô hình nổi tiếng của Google.
        max_length=512: Độ dài tối đa của mỗi đoạn (tính bằng số lượng token).
        stride=256: Đây là bước nhảy của "cửa sổ".
            Nếu stride nhỏ hơn max_length, các đoạn sẽ gối đầu (overlap) lên nhau.
            Trong ví dụ này: Đoạn 1 lấy từ token 0 đến 512. Đoạn 2 sẽ nhảy 256 bước, tức là lấy từ token 256 đến 768. Vậy chúng sẽ gối đầu lên nhau 256 token.
    2. Quy trình hoạt động của hàm chunk_text:
        input_ids = tokenizer(...): Chuyển toàn bộ văn bản đầu vào thành một dãy số dài (các Token ID).
        Vòng lặp while:
            Nó cắt một đoạn ID từ vị trí i đến i + max_length.
            tokenizer.decode(...): Chuyển đoạn ID đó ngược lại thành văn bản người đọc được.
            i += stride: Nhảy "cửa sổ" đi một đoạn bằng giá trị stride.
        Kết quả: Trả về một danh sách các chuỗi văn bản đã được cắt theo đúng giới hạn token của mô hình.
    3. Tại sao cách này lại quan trọng trong RAG?
        Chính xác về mặt kỹ thuật: Các mô hình AI như GPT, T5, Llama đều có giới hạn "Context Window" (ví dụ: 4096 tokens). Nếu bạn cắt theo ký tự, bạn có thể vô tình nạp quá nhiều token làm mô hình bị lỗi. Cắt theo token giúp bạn kiểm soát tuyệt đối dung lượng đầu vào.
        Giữ ngữ cảnh cực tốt: Nhờ có stride (overlap), đoạn sau luôn chứa một phần thông tin của đoạn trước. Điều này giúp ngăn chặn việc một thông tin quan trọng bị cắt làm đôi ở giữa hai đoạn, khiến AI không hiểu được toàn bộ ý nghĩa.
    4. Lưu ý về đoạn code:
    Trong đoạn code bạn cung cấp, dòng:
    model = AutoModelForSeq2SeqLM.from_pretrained("t5-base")
    thực tế không được sử dụng trong hàm chia nhỏ văn bản. Việc tải mô hình (model) này sẽ tốn khá nhiều RAM/VRAM. Nếu bạn chỉ cần chia nhỏ văn bản, bạn chỉ cần dùng tokenizer là đủ.
    Tóm lại: Đây là kỹ thuật Sliding Window Tokenization. Nó đảm bảo các đoạn văn bản có độ dài hoàn hảo cho AI và không làm mất thông tin quan trọng tại các điểm cắt.

```python

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load the tokenizer and model
tokenizer = AutoTokenizer.from_pretrained("t5-base")

def chunk_text(text, max_length=512, stride=256):

    input_ids = tokenizer(text, return_tensors="pt").input_ids[0]

    chunks = []
    i = 0
    while i < len(input_ids):
        end_idx = min(i + max_length, len(input_ids))
        chunk_ids = input_ids[i:end_idx]
        chunks.append(tokenizer.decode(chunk_ids, skip_special_tokens=True))
        i += stride

    return chunks

# Example usage
text = "Astronauts travel through space. The journey is difficult. Exploration of Mars is a recent milestone in space science."
chunks = chunk_text(text)

for chunk in chunks:
  print(chunk)
```

    8. Content-Aware Chunking
    This method adapts chunking based on content characteristics (e.g., chunking text at paragraph level, tables as separate entities).

    When to Use:
    For documents with heterogeneous content, such as eBooks or technical manuals, chunking must vary based on content type.

    Đây là phương pháp Content-Aware Chunking (Chia nhỏ dựa trên cấu trúc nội dung). Thay vì dựa vào độ dài ký tự hay số lượng từ, cách này sử dụng các dấu hiệu logic (như tiêu đề, từ khóa quan trọng) để phân tách văn bản.
    Dưới đây là giải thích chi tiết:
    1. Cách thức hoạt động của hàm:
    Hàm này quét qua từng dòng của văn bản và tìm kiếm các "điểm neo" (anchors):
        Dấu hiệu tiêu đề: Các dòng bắt đầu bằng ## hoặc ### (định dạng Markdown cho tiêu đề cấp 2, cấp 3).
        Từ khóa logic: Các dòng bắt đầu bằng Introduction (Mở đầu) hoặc Conclusion (Kết luận).
    Logic xử lý:
        Nếu gặp một dòng là "tiêu đề" (marker), nó sẽ đóng đoạn (chunk) hiện tại lại và bắt đầu một đoạn mới.
        Tiêu đề đó sẽ trở thành dòng đầu tiên của đoạn mới.
        Tất cả các dòng không phải tiêu đề sẽ được gom vào đoạn hiện tại.
    2. Ưu điểm vượt trội:
        Giữ trọn vẹn ngữ nghĩa: Đây là cách chia nhỏ "thông minh" nhất về mặt con người. Nó đảm bảo rằng một chủ đề (nằm dưới một tiêu đề) sẽ không bị xé lẻ ra. Toàn bộ nội dung của phần "Introduction" sẽ nằm chung một chỗ.
        Tối ưu cho truy xuất (Retrieval): Khi người dùng hỏi về một chủ đề cụ thể, hệ thống RAG sẽ tìm thấy chính xác đoạn văn bản bắt đầu bằng tiêu đề đó, giúp AI trả lời cực kỳ chính xác.
    3. Nhược điểm cần lưu ý:
        Kích thước không đồng đều: Nếu một chương trong tài liệu của bạn quá dài (ví dụ 5000 chữ), hàm này vẫn sẽ gom nó thành 1 chunk duy nhất. Điều này có thể vượt quá giới hạn (Context Window) của các mô hình AI.
        Phụ thuộc vào định dạng: Nếu văn bản không có tiêu đề rõ ràng hoặc dùng định dạng khác (không phải ##), hàm này sẽ không hoạt động hiệu quả.
    4. Ứng dụng thực tế:
    Cách tiếp cận này rất phổ biến khi bạn làm việc với:
        Tài liệu kỹ thuật: Chia theo các mục lục.
        Văn bản luật: Chia theo các "Điều", "Khoản".
        Bài báo khoa học: Chia theo "Abstract", "Methodology", "Result".

```python
sample_text = """
Introduction

Data Science is an interdisciplinary field that uses scientific methods, processes,
 algorithms, and systems to extract knowledge and insights from structured and
 unstructured data. It draws from statistics, computer science, machine learning,
 and various data analysis techniques to discover patterns, make predictions, and
 derive actionable insights.

Data Science can be applied across many industries, including healthcare, finance,
 marketing, and education, where it helps organizations make data-driven decisions,
  optimize processes, and understand customer behaviors.

Overview of Big Data

Big data refers to large, diverse sets of information that grow at ever-increasing
rates. It encompasses the volume of information, the velocity or speed at which it
is created and collected, and the variety or scope of the data points being
covered.

Data Science Methods

There are several important methods used in Data Science:

1. Regression Analysis
2. Classification
3. Clustering
4. Neural Networks

Challenges in Data Science

- Data Quality: Poor data quality can lead to incorrect conclusions.
- Data Privacy: Ensuring the privacy of sensitive information.
- Scalability: Handling massive datasets efficiently.

Conclusion

Data Science continues to be a driving force in many industries, offering insights
that can lead to better decisions and optimized outcomes. It remains an evolving
field that incorporates the latest technological advancements.
"""


def content_aware_chunk(text):
    chunks = []
    current_chunk = []
    for line in text.splitlines():
        if line.startswith(('##', '###', 'Introduction', 'Conclusion')):
            if current_chunk:
                chunks.append('\n'.join(current_chunk))
            current_chunk = [line]
        else:
            current_chunk.append(line)
    if current_chunk:
        chunks.append('\n'.join(current_chunk))
    return chunks

# Applying Content-Aware Chunking
content_chunks = content_aware_chunk(sample_text)
for chunk in content_chunks:
    print(chunk, '\n---\n')
```

    9. Token-Based Chunking
        Token-based chunking splits text based on a fixed number of tokens rather than words or sentences. It uses tokenizers from NLP models (e.g., Hugging Face’s transformers).

    When to Use:
        For models that operate on tokens, such as transformer-based models with token limits (e.g., GPT-3 or GPT-4).

    Advantages:
        Works well with transformer-based models.
        Ensures token limits are respected.

    Disadvantages:
        Tokenization may split sentences or break context.
        Not always aligned with natural language boundaries.


    Đây là phương pháp Token-based Chunking (Chia nhỏ dựa trên đơn vị Token) sử dụng bộ mã hóa của mô hình GPT-2.
    Về cơ bản, nó giống với kỹ thuật chia theo Token mà bạn đã xem ở ví dụ T5, nhưng được viết theo cách tối giản hơn và sử dụng GPT-2 Tokenizer.
    1. Giải thích chi tiết từng dòng:
        tokenizer = GPT2Tokenizer.from_pretrained("gpt2"):
            Tải bộ mã hóa của GPT-2. Dù GPT-2 là mô hình cũ, nhưng bộ Tokenizer của nó vẫn rất phổ biến để minh họa cách AI "nhìn" văn bản.
        tokens = tokenizer(text)["input_ids"]:
            Chuyển toàn bộ văn bản đầu vào thành một danh sách các con số (Token IDs).
        chunks = [tokens[i:i + max_tokens] for i in range(0, len(tokens), max_tokens)]:
            Sử dụng list comprehension để chia danh sách các Token ID thành các phần nhỏ, mỗi phần có đúng max_tokens (mặc định là 200).
            Lưu ý: Cách này cắt "phập" một phát cứ mỗi 200 token mà không có gối đầu (overlap).
        tokenizer.decode(chunk):
            Chuyển các con số ngược lại thành văn bản mà con người đọc được để trả về kết quả.

```python
from transformers import GPT2Tokenizer

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

def token_based_chunk(text, max_tokens=200):
    tokens = tokenizer(text)["input_ids"]
    chunks = [tokens[i:i + max_tokens] for i in range(0, len(tokens), max_tokens)]
    return [tokenizer.decode(chunk) for chunk in chunks]

# Applying Token-Based Chunking
token_chunks = token_based_chunk(sample_text)
for chunk in token_chunks:
    print(chunk, '\n---\n')
```

    10. Topic-Based Chunking
        This strategy splits the document based on topics using techniques like Latent Dirichlet Allocation (LDA) or other topic modeling algorithms to segment the text.

    When to Use:
        For documents that cover multiple topics, such as news articles, research papers, or reports with diverse subject matter.

    Advantages:
        Groups related information together.
        Helps in focused retrieval based on specific topics.

    Disadvantages:
        Requires additional processing (topic modeling).
        May not be precise for short documents or overlapping topics.


    Đây là phương pháp Topic-based Chunking (Chia nhỏ dựa trên chủ đề) sử dụng thuật toán LDA (Latent Dirichlet Allocation).
    Thay vì chia theo ngữ nghĩa (K-Means) hay cấu trúc (Tiêu đề), cách này sử dụng xác suất thống kê để tìm ra các "chủ đề ẩn" bên trong văn bản và nhóm các câu lại theo các chủ đề đó.
    1. Giải thích các thành phần kỹ thuật:
        CountVectorizer: Chuyển đổi các câu thành một "ma trận đếm từ" (Bag of Words). Nó đếm xem mỗi từ xuất hiện bao nhiêu lần trong mỗi câu.
        LatentDirichletAllocation (LDA): Một thuật toán mô hình hóa chủ đề (Topic Modeling). Nó giả định rằng một văn bản là sự pha trộn của nhiều chủ đề, và mỗi chủ đề là sự pha trộn của nhiều từ.
        n_components=3: Bạn yêu cầu thuật toán tìm ra 3 chủ đề chính trong văn bản.
    2. Quy trình hoạt động của hàm:
        Tách câu: Chia văn bản thành danh sách các câu đơn lẻ dựa trên dấu chấm (.).
        Vector hóa: Biến các câu thành dạng số mà thuật toán LDA có thể hiểu được.
        Huấn luyện LDA: Thuật toán quét qua toàn bộ các câu để tìm ra 3 nhóm từ thường xuyên xuất hiện cùng nhau (đại diện cho 3 chủ đề).
        Trích xuất từ khóa: Với mỗi chủ đề, nó lấy ra 5 từ quan trọng nhất để tạo thành nhãn (ví dụ: "Topic 1: space, astronauts, Mars").
        Gán chủ đề cho từng câu:
            Với mỗi câu, thuật toán tính toán xem nó thuộc về chủ đề nào nhất (xác suất cao nhất).
            Trả về kết quả là cặp: (Tên chủ đề, Nội dung câu).
    3. Ưu điểm và Nhược điểm:
        Ưu điểm:
            Phát hiện nội dung ẩn: Rất mạnh khi xử lý các văn bản dài và hỗn hợp nhiều nội dung (ví dụ: một biên bản họp thảo luận về cả "Tài chính", "Nhân sự" và "Kế hoạch sản xuất").
            Nhãn có ý nghĩa: Nó không chỉ chia nhóm mà còn cho bạn biết nhóm đó nói về cái gì (thông qua các từ khóa).
        Nhược điểm:
            Cần dữ liệu đủ lớn: LDA hoạt động không tốt trên các văn bản quá ngắn vì không đủ dữ liệu để thống kê xác suất từ.
            Số lượng chủ đề (num_topics): Bạn phải tự đoán xem văn bản có bao nhiêu chủ đề. Nếu chọn sai (ví dụ văn bản có 5 chủ đề mà bạn chọn 3), kết quả sẽ bị trộn lẫn.
            Thứ tự câu: Cách này có thể nhóm câu số 1 và câu số 100 lại với nhau nếu chúng cùng chủ đề, làm mất đi tính tuần tự của văn bản.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import numpy as np

def topic_based_chunk(text, num_topics=3):
    # Split the text into sentences for chunking
    sentences = text.split('. ')

    # Vectorize the sentences
    vectorizer = CountVectorizer()
    sentence_vectors = vectorizer.fit_transform(sentences)

    # Apply LDA for topic modeling
    lda = LatentDirichletAllocation(n_components=num_topics, random_state=42)
    lda.fit(sentence_vectors)

    # Get the topic-word distribution
    topic_word = lda.components_
    vocabulary = vectorizer.get_feature_names_out()

    # Identify the top words for each topic
    topics = []
    for topic_idx, topic in enumerate(topic_word):
        top_words_idx = topic.argsort()[:-6:-1]
        topic_keywords = [vocabulary[i] for i in top_words_idx]
        topics.append("Topic {}: {}".format(topic_idx + 1, ', '.join(topic_keywords)))

    # Generate chunks with topics
    chunks_with_topics = []
    for i, sentence in enumerate(sentences):
        topic_assignments = lda.transform(vectorizer.transform([sentence]))
        assigned_topic = np.argmax(topic_assignments)
        chunks_with_topics.append((topics[assigned_topic], sentence))

    return chunks_with_topics


# Get topic-based chunks
topic_chunks = topic_based_chunk(sample_text, num_topics=3)

# Display results
for topic, chunk in topic_chunks:
    print(f"{topic}: {chunk}\n")
```

    11. Keyword-Based Chunking
        This method chunks documents based on predefined keywords or phrases that signal topic shifts (e.g., “Introduction,” “Conclusion”).

    When to Use:
        Best for documents that follow a clear structure, such as scientific papers or technical specifications.

    Advantages:
        Captures natural topic breaks based on keywords.
        Works well for structured documents.

    Disadvantages:
        Requires a predefined set of keywords.
        Not adaptable to unstructured text.


    Đây là phương pháp Keyword-based Chunking (Chia nhỏ dựa trên từ khóa). Nó là một biến thể của Content-Aware Chunking, nhưng thay vì dựa vào định dạng (như dấu ##), nó dựa vào các từ vựng cụ thể mà bạn quy định trước.
    Dưới đây là giải thích ngắn gọn:
    1. Cách thức hoạt động:
    Hàm này quét qua từng dòng của văn bản và đóng vai trò như một "người gác cổng":
        Danh sách từ khóa: Bạn cung cấp một danh sách các từ quan trọng thường dùng làm tiêu đề mục như Introduction, Methods, Conclusion...
        Điều kiện ngắt: Nếu một dòng chứa bất kỳ từ nào trong danh sách keywords, hàm sẽ hiểu đó là điểm bắt đầu của một phần mới.
        Gom nhóm: Nó sẽ đóng đoạn hiện tại lại và bắt đầu gom các dòng tiếp theo vào một đoạn mới cho đến khi gặp một từ khóa khác.
    2. Sự khác biệt với Content-Aware Chunking:
        Content-Aware (bài trước): Thường tìm kiếm các ký hiệu cấu trúc (như ## trong Markdown). Nó dựa vào hình thức trình bày.
        Keyword-based (bài này): Tìm kiếm nội dung chữ cụ thể. Nó dựa vào ngữ nghĩa của tiêu đề.
            Ví dụ: Nếu văn bản của bạn không dùng dấu ## mà chỉ viết chữ "Conclusion" in đậm, thì cách dùng từ khóa này sẽ hiệu quả hơn.
    3. Ưu điểm và Nhược điểm:
        Ưu điểm:
            Tùy biến cao: Bạn có thể chỉ định chính xác những phần nào bạn muốn tách ra (ví dụ: chỉ tách phần "Challenges" để phân tích rủi ro).
            Phù hợp với tài liệu chuẩn hóa: Rất hữu ích cho các loại báo cáo, hợp đồng, hoặc hồ sơ y tế có các mục lục luôn cố định.
        Nhược điểm:
            Dễ bị ngắt nhầm (False Positives): Nếu từ "Conclusion" xuất hiện ở giữa một câu văn bình thường (không phải tiêu đề), hàm này vẫn sẽ cắt văn bản tại đó, làm hỏng mạch ý nghĩa.
            Nhạy cảm với chữ hoa/thường: Nếu danh sách từ khóa là Conclusion nhưng văn bản ghi là CONCLUSION, hàm có thể sẽ bỏ sót (trừ khi bạn thêm logic .lower()).
    4. Ứng dụng trong RAG:
    Phương pháp này thường được dùng làm bộ lọc sơ bộ.
        Ví dụ: Bạn có hàng ngàn bài nghiên cứu khoa học. Bạn chỉ muốn lấy phần "Methods" để nạp vào AI nhằm so sánh các phương pháp nghiên cứu với nhau. Bạn dùng hàm này với từ khóa Methods để trích xuất đúng đoạn đó.

```python
def keyword_based_chunk(text, keywords):
    chunks = []
    current_chunk = []
    for line in text.splitlines():
        if any(keyword in line for keyword in keywords):
            if current_chunk:
                chunks.append('\n'.join(current_chunk))
            current_chunk = [line]
        else:
            current_chunk.append(line)
    if current_chunk:
        chunks.append('\n'.join(current_chunk))
    return chunks

# Applying Keyword-Based Chunking
keywords = ["Introduction", "Overview", "Conclusion", "Methods", "Challenges"]
keyword_chunks = keyword_based_chunk(sample_text, keywords)
for chunk in keyword_chunks:
    print(chunk, '\n---\n')
```

    Trade-offs Between Chunk Size, Retrieval Speed, and Accuracy
    Larger Chunks:
        Pros: More context, better accuracy.
        Cons: Slower retrieval, higher memory usage.

    Smaller Chunks:
        Pros: Faster retrieval, lower memory usage.
        Cons: Less context, potential loss of accuracy.


    Optimization Tactics:
    Sliding Window Chunking: Combines the best of both worlds:
        Preserves context by overlapping chunks.
        Maintains reasonable retrieval speed.
    Token-Based Chunking:
        Ensures chunks fit within model token limits.
        Maintains efficient retrieval.
