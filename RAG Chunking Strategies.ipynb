    Small Chunk Size
        Advantages
            Fine-grained search with higher accuracy for specific queries.
            Useful for datasets with highly distributed information.
        Disadvantages
            Context fragmentation: important context might span multiple chunks.
            Increased resource consumption (storage and compute).

    Large Chunk Size
        Advantages
            Better for context-aware retrieval.
            Reduces embedding computation and storage.
        Disadvantages
            Low recall for specific keyword-based queries.
            Risks of irrelevant content being retrieved.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

# Sample document (larger for better testing)
document = """
Elon Musk is the CEO of Tesla. Tesla's mission is to accelerate the world's transition to sustainable energy.
Tesla manufactures electric vehicles (EVs), battery energy storage, solar panels, and related products. SpaceX, another company led by Musk,
aims to make space travel accessible to humanity by developing reusable rockets.
Musk has also co-founded Neuralink, which focuses on connecting the human brain to computers using advanced neural interfaces.
Recently, Musk has expressed his concerns about artificial intelligence, advocating for regulation to prevent potential harm.
Musk believes that governments should play an active role in ensuring the safe development of AI technologies.
His innovative ventures span automotive, space exploration, energy, AI, and neuroscience, making him a pivotal figure in modern technology.
"""

# chunk_size=10: Mỗi đoạn văn bản (chunk) sau khi cắt sẽ có độ dài tối đa là 10 ký tự.
# chunk_overlap=2: Đoạn sau sẽ lặp lại (gối đầu) 2 ký tự cuối của đoạn trước. Việc này giúp giữ mạch văn, đảm bảo ngữ cảnh không bị cắt đứt đột ngột giữa các đoạn.
# RecursiveCharacterTextSplitter: Đây là bộ cắt thông minh, nó sẽ cố gắng không cắt ngang giữa từ hoặc câu (ưu tiên cắt tại khoảng trắng hoặc dấu dòng) cho đến khi đạt gần mức 10 ký tự.
# split_text(document): Thực hiện hành động cắt nội dung của biến document và lưu kết quả vào danh sách small_chunk.
small_splitter = RecursiveCharacterTextSplitter(chunk_size=10, chunk_overlap=2)
small_chunk = small_splitter.split_text(document)
print("Smaller chunks -")
print(small_chunk)

#large chunk size
large_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=10)
large_chunk = large_splitter.split_text(document)
print("Larger chunks -")
print(large_chunk)

# Initialize embedding model
# embedding_model: Là "bộ thông dịch" biến câu chữ thành các dãy số (Vector). Máy tính không hiểu ý nghĩa của từ, nhưng nó có thể so sánh khoảng cách giữa các dãy số để biết hai câu có ý nghĩa gần nhau hay không.
# all-MiniLM-L6-v2: Một mô hình cực kỳ phổ biến, nhỏ gọn và chạy rất nhanh trên máy tính cá nhân (local). Nó biến mỗi đoạn văn bản thành một vector có 384 chiều.
embedding_model = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Create vector stores
# Lấy dữ liệu: Nhận vào danh sách các đoạn văn bản (small_chunk hoặc large_chunk).
# Tính toán: Dùng embedding_model để tính toán Vector cho từng đoạn văn bản đó.
# Lưu trữ: Lưu cả văn bản gốc và Vector tương ứng của nó vào database Chroma.
small_vector_store = Chroma.from_texts(small_chunk, embedding_model)
large_vector_store = Chroma.from_texts(large_chunk, embedding_model)

# Semantic query
query = "What are Musk's views on artificial intelligence and government regulations?"

# Perform search
small_results = small_vector_store.similarity_search(query, k=1)
large_results = large_vector_store.similarity_search(query, k=1)

# câu trả lời khá giống nhau vì dữ liệu nhỏ, nếu dữ liệu lớn hơn có thể khác

# Display results
print("-----------------------------------")
print("Results")
print("-----------------------------------")
print("Small Chunk Results:", small_results)
print("Large Chunk Results:", large_results)
```

    1. Fixed-size Chunking (Sliding Window)
    What It Is:
    Fixed-size chunking divides the text into fixed-length pieces or windows. This is done by splitting the text into chunks of a predetermined size (e.g., 512 tokens). This strategy is simple but can be inefficient for longer texts, especially if chunks cut across meaningful sections (like sentences).

    Use Case:
    Best for datasets where text length is fairly uniform or where processing uniform chunks is necessary.

```python
#words = text.split():
# Cắt chuỗi văn bản thành một danh sách (list) các từ đơn lẻ.
# Ví dụ: "Hello world" thành ['Hello', 'world']. Mặc định nó sẽ cắt dựa trên khoảng trắng.
# chunks = [words[i:i + chunk_size] for i in range(0, len(words), chunk_size)]:
# Đây là một kỹ thuật "chia mảng thành các phần bằng nhau" (array chunking).
# Nó duyệt qua danh sách từ, mỗi lần nhảy một bước bằng chunk_size.
# Kết quả là một danh sách chứa các danh sách con, mỗi danh sách con có tối đa 5 từ (theo ví dụ).
# return [' '.join(chunk) for chunk in chunks]:
# Gộp các từ trong mỗi danh sách con lại thành một câu hoàn chỉnh bằng dấu cách.
def fixed_size_chunk(text, chunk_size=512):
    # Split text into words and chunk them into fixed-size parts
    words = text.split()
    chunks = [words[i:i + chunk_size] for i in range(0, len(words), chunk_size)]
    print(chunks)
    return [' '.join(chunk) for chunk in chunks]

# Example text (typically a long document)
text = "This is a very long text that should be chunked into smaller pieces for efficient processing."

# Chunk into 5 words per chunk
chunks = fixed_size_chunk(text, chunk_size=8)
# print(chunks)

# Display the chunks
for i, chunk in enumerate(chunks):
    print(f"Chunk {i + 1}: {chunk}")
```
