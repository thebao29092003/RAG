Vector Embeddings


    Vector Embeddings là một trong những khái niệm quan trọng nhất của trí tuệ nhân tạo hiện đại. Nếu không có nó, các mô hình như ChatGPT, Midjourney hay các hệ thống tìm kiếm thông minh sẽ không thể hoạt động.
    Dưới đây là cái nhìn tổng quan dễ hiểu nhất về Vector Embeddings:
    1. Định nghĩa: "Số hóa" ý nghĩa
    Máy tính không hiểu từ ngữ, hình ảnh hay âm thanh như con người. Nó chỉ hiểu những con số.
    Vector Embedding là quá trình chuyển đổi một đối tượng dữ liệu (một từ, một câu, một bức ảnh) thành một dãy số thực (gọi là một Vector).
        Ví dụ: Thay vì hiểu chữ "Mèo", máy tính sẽ thấy một dãy số như: [0.12, -0.54, 0.89, ...]
    2. Nguyên lý cốt lõi: Khoảng cách = Ý nghĩa
    Điểm kỳ diệu của Embeddings là: Những thứ có ý nghĩa giống nhau sẽ được chuyển thành những dãy số gần giống nhau.
    Hãy tưởng tượng một không gian 3 chiều:
        Từ "Chó" và "Mèo" sẽ có tọa độ nằm rất gần nhau vì chúng đều là thú cưng.
        Từ "Cái bàn" sẽ nằm ở một góc rất xa hai từ trên vì nó là đồ nội thất.
        Từ "Apple" (quả táo) sẽ nằm gần từ "Chuối", nhưng nếu trong ngữ cảnh công nghệ, nó sẽ dịch chuyển về phía từ "iPhone".
    3. Cách hoạt động: Không gian đa chiều
    Trong thực tế, các Vector Embedding không chỉ có 2 hay 3 chiều mà có tới hàng trăm hoặc hàng nghìn chiều (ví dụ: 768 chiều, 1536 chiều). Mỗi chiều đại diện cho một đặc tính ẩn mà mô hình AI tự học được:
        Chiều 1: Có phải là động vật không?
        Chiều 2: Có tính chất hoàng gia không?
        Chiều 3: Có dùng trong công nghệ không?
    Phép toán nổi tiếng:
    Một ví dụ kinh điển để minh họa tính logic của không gian vector:
    Vector("Vua") - Vector("Đàn ông") + Vector("Đàn bà") = Vector("Nữ hoàng")
    4. Các loại Embedding phổ biến
        Word Embeddings: Chuyển từng từ thành vector (Ví dụ: Word2Vec, GloVe).
        Sentence/Document Embeddings: Chuyển cả câu hoặc đoạn văn thành vector (Đây là cái bạn dùng trong RAG với HuggingFaceEmbeddings).
        Image Embeddings: Chuyển hình ảnh thành vector để tìm kiếm ảnh tương đồng (Ví dụ: CLIP của OpenAI).
    5. Tại sao Vector Embeddings lại quan trọng?
        Tìm kiếm theo ý nghĩa (Semantic Search): Bạn tìm "phương tiện giao thông", máy tính có thể trả về "xe máy", "ô tô" dù trong văn bản không hề có từ "phương tiện".
        Hệ thống gợi ý (Recommendation Systems): TikTok hay Netflix biết bạn thích gì vì chúng so sánh Vector sở thích của bạn với Vector của các video/phim.
        RAG (Retrieval-Augmented Generation):
            Bạn chuyển toàn bộ tài liệu thành Vector và lưu vào Vector Database (như ChromaDB).
            Khi có câu hỏi, bạn chuyển câu hỏi thành Vector.
            Máy tính so sánh Vector câu hỏi với kho Vector tài liệu để tìm ra đoạn văn phù hợp nhất.
    6. Cách tạo ra Vector Embeddings?
    Bạn không tự tay viết ra các con số này. Bạn sử dụng các Embedding Models đã được huấn luyện sẵn trên hàng tỷ dữ liệu:
        OpenAI: text-embedding-3-small (Trả phí).
        Hugging Face: all-MiniLM-L6-v2, BGE-m3 (Miễn phí, chạy local).
        Google: Vertex AI Embeddings.


    Image Embeddings
    We use a pre-trained vision transformer (ViT) model from hugging Face for image embeddings


    Đây là ví dụ về Image Embeddings sử dụng mô hình ViT (Vision Transformer).
    Nếu các ví dụ trước giúp máy tính "hiểu" văn bản, thì đoạn code này giúp máy tính "nhìn" và chuyển hóa hình ảnh thành một dãy số (Vector) đại diện cho nội dung của bức ảnh đó.
    Giải thích chi tiết từng phần:
    1. Khởi tạo mô hình ViT:
        ViTImageProcessor: Giống như bộ tiền xử lý. Nó sẽ lấy ảnh gốc và biến đổi về đúng chuẩn mà AI yêu cầu (ví dụ: đổi kích thước về 224x224 pixels, chuẩn hóa màu sắc).
        ViTModel: Mô hình Vision Transformer của Google. Nó không dùng các lớp cuộn (CNN) truyền thống mà chia ảnh thành các ô vuông nhỏ (patches) và dùng cơ chế Attention (giống GPT) để hiểu mối quan hệ giữa các phần của ảnh.
        vit-base-patch16-224-in21k:
            base: Kích thước mô hình trung bình.
            patch16: Chia ảnh thành các ô 16x16.
            224: Đầu vào ảnh là 224x224.
    2. Tiền xử lý ảnh:
    inputs = feature_extractor(images=image, return_tensors="pt")
        Biến bức ảnh từ file .jpg thành một Tensor (ma trận số) để PyTorch có thể tính toán.
    3. Tạo Embedding (Trái tim của đoạn code):
    with torch.no_grad():
        outputs = model(**inputs)
        image_embedding = outputs.last_hidden_state.mean(dim=1)
            outputs.last_hidden_state: Khi ảnh đi qua mô hình, kết quả cuối cùng là một tập hợp các vector cho từng ô vuông (patch) của ảnh.
            mean(dim=1) (Average Pooling): Vì ta có nhiều vector cho nhiều ô vuông, ta lấy trung bình cộng của chúng để tạo ra duy nhất một vector đại diện cho toàn bộ ý nghĩa của bức ảnh.
            torch.no_grad(): Tắt chế độ tính đạo hàm để tiết kiệm RAM và chạy nhanh hơn (vì ta chỉ đang lấy dữ liệu, không huấn luyện lại AI).

    Kết quả đầu ra:
        shape: Thường sẽ là [1, 768]. Nghĩa là 1 bức ảnh được đại diện bởi một dãy gồm 768 con số.
        image_embedding: Chính là dãy số "vân tay" của bức ảnh.


```python
from transformers import ViTImageProcessor, ViTModel
from PIL import Image
import torch

# Load pre-trained ViT model and feature extractor
feature_extractor = ViTImageProcessor.from_pretrained(
    'google/vit-base-patch16-224-in21k'
)
model = ViTModel.from_pretrained(
    'google/vit-base-patch16-224-in21k'
)

# Load and preprocess image
image = Image.open("example.png")
inputs = feature_extractor(images=image, return_tensors="pt")

# Generate embeddings
with torch.no_grad():
    outputs = model(**inputs)
    image_embedding = outputs.last_hidden_state.mean(dim=1)  # Average pooling

print(image_embedding.shape)  # Embedding shape
print(image_embedding)
```

    Embedding Audio
    We use a pre-trained Wav2Vec2 model for audio embeddings

```python
#Cần numpy 2.3

from transformers import Wav2Vec2FeatureExtractor, Wav2Vec2Model
import torch
import librosa

# Load pre-trained Wav2Vec2 model and feature extractor
feature_extractor = Wav2Vec2FeatureExtractor.from_pretrained(
    "facebook/wav2vec2-base-960h"
)
model = Wav2Vec2Model.from_pretrained(
    "facebook/wav2vec2-base-960h"
)

# Load and preprocess audio
audio_file = "example.mp3"  # Replace with your audio file
audio, rate = librosa.load(audio_file, sr=16000)  # Resample to 16 kHz
inputs = feature_extractor(
    audio,
    return_tensors="pt",
    sampling_rate=rate
)

# Generate embeddings
with torch.no_grad():
    outputs = model(**inputs)
    audio_embedding = outputs.last_hidden_state.mean(dim=1)  # Average pooling

print(audio_embedding.shape)
print(audio_embedding)
```

    Embedding Text
    we use a pre-trained BERT model for text embeddings


    Đây là một ví dụ kinh điển về cách tạo Text Embeddings (Vectơ hóa văn bản) sử dụng mô hình BERT — một trong những mô hình AI mang tính cách mạng nhất trong việc hiểu ngôn ngữ tự nhiên.
    Nếu các ví dụ trước bạn dùng thư viện LangChain để làm tự động, thì đây là cách làm "thủ công" (low-level) để bạn hiểu thực sự bên trong "bộ não" của AI hoạt động như thế nào.
    Giải thích chi tiết từng bước:
    1. Khởi tạo mô hình và bộ mã hóa (Tokenizer):
        bert-base-uncased: Đây là mô hình BERT phiên bản cơ bản. "Uncased" nghĩa là nó không phân biệt chữ hoa chữ thường (ví dụ: "Apple" và "apple" được coi là như nhau).
        AutoTokenizer: Chuyển văn bản thành các con số (Token IDs) mà mô hình BERT có thể đọc được.
        AutoModel: Chính là "bộ não" chứa hàng triệu tham số đã được huấn luyện để hiểu ngữ nghĩa của từ ngữ.
    2. Tiền xử lý văn bản (Tokenization):
    inputs = tokenizer(
        text,
        return_tensors="pt", # Trả về định dạng PyTorch tensor
        padding=True,        # Tự động thêm các số 0 nếu câu quá ngắn (để đủ độ dài chuẩn)
        truncation=True      # Tự động cắt bớt nếu câu quá dài (quá giới hạn 512 tokens)
    )
    3. Tạo Embedding (Quá trình suy luận):
        with torch.no_grad(): Tắt chế độ tính đạo hàm (gradient). Vì chúng ta chỉ đang dùng mô hình để lấy dữ liệu, không phải để huấn luyện nó, việc này giúp tiết kiệm rất nhiều RAM và chạy nhanh hơn.
        outputs = model(**inputs): Đưa các con số vào mô hình. Kết quả trả về là một ma trận số phức tạp.
    4. Trích xuất ý nghĩa (Mean Pooling):
        text_embedding = outputs.last_hidden_state.mean(dim=1)
        last_hidden_state: BERT trả về một vectơ cho từng token trong câu. Nếu câu có 10 từ, bạn sẽ có 10 vectơ.
        mean(dim=1): Để đại diện cho cả câu bằng một vectơ duy nhất, chúng ta lấy trung bình cộng của tất cả các vectơ từ đó. Đây gọi là kỹ thuật Mean Pooling.
    Kết quả đầu ra:
        shape: Sẽ là torch.Size([1, 768]).
        1: Số lượng câu đầu vào.
        768: Số chiều của không gian vectơ. Nghĩa là BERT đã mô tả ý nghĩa câu văn của bạn bằng 768 con số thực.
        text_embedding: Chính là "vân tay số" của câu văn đó.

```python
from transformers import AutoTokenizer, AutoModel
import torch

# Load pre-trained BERT model and tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

# Input text
text = "This is an example sentence for embedding."
inputs = tokenizer(
    text,
    return_tensors="pt",
    padding=True,
    truncation=True
)

# Generate embeddings
with torch.no_grad():
    outputs = model(**inputs)
    text_embedding = outputs.last_hidden_state.mean(dim=1)  # Average pooling

print(text_embedding.shape)  # Embedding shape
print(text_embedding)
```
