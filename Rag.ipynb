RAG


    RAG (Retrieval-Augmented Generation) hay Thế hệ tăng cường truy xuất là một kỹ thuật giúp các mô hình ngôn ngữ lớn (LLM) như GPT-4, Claude... trở nên thông minh và chính xác hơn bằng cách kết nối chúng với các nguồn dữ liệu tin cậy bên ngoài (của riêng bạn hoặc tổ chức của bạn).
    Nói một cách đơn giản nhất: LLM là một đầu bếp giỏi, còn RAG là cuốn sách dạy nấu ăn và kho nguyên liệu tươi mới nhất mà đầu bếp đó chưa từng được học qua.
    1. Tại sao cần RAG? (Vấn đề của LLM thuần túy)
    Dù LLM rất mạnh, chúng có 3 nhược điểm lớn:
        Kiến thức cũ (Knowledge Cut-off): LLM chỉ biết những gì có trong dữ liệu huấn luyện (ví dụ: GPT-4 có thể chỉ biết dữ liệu đến năm 2023).
        Ảo giác (Hallucination): Khi không biết câu trả lời, LLM có xu hướng "bịa" ra thông tin trông có vẻ rất thuyết phục.
        Không biết dữ liệu nội bộ: LLM không biết gì về các file PDF, email, hay database bán hàng riêng tư của công ty bạn.
    RAG giải quyết cả 3 vấn đề này.
    2. Mô hình hoạt động của RAG (Quy trình 3 bước)
    Quy trình RAG diễn ra như sau khi bạn đặt một câu hỏi:
        Retrieval (Truy xuất): Thay vì gửi thẳng câu hỏi cho LLM, hệ thống sẽ đi "lục tìm" trong kho dữ liệu của bạn (PDF, Database, Wiki) để tìm ra những đoạn thông tin có liên quan nhất đến câu hỏi đó.
        Augmentation (Tăng cường): Hệ thống lấy những đoạn thông tin tìm được, "dán" chúng vào cùng với câu hỏi ban đầu để tạo thành một yêu cầu (Prompt) khổng lồ và đầy đủ thông tin.
        Generation (Tạo câu trả lời): LLM nhận được Prompt có sẵn kiến thức này. Nó không cần nhớ lại kiến thức cũ nữa mà chỉ cần đọc những gì bạn cung cấp và tổng hợp thành câu trả lời chính xác.
    3. Các thành phần kỹ thuật trong một hệ thống RAG
        Để xây dựng RAG, bạn cần các thành phần sau (đây là lúc kỹ năng Data Engineering của bạn phát huy tác dụng):
        Loaders: Các công cụ để đọc dữ liệu từ nhiều nguồn (PDF, Excel, API GitHub, SQL...).
        Chunking (Chia nhỏ): Cắt văn bản dài thành các đoạn nhỏ (ví dụ mỗi đoạn 500 chữ) để dễ tìm kiếm.
        Embedding Model: Một mô hình AI chuyên biệt dùng để biến các đoạn văn bản thành dãy số (gọi là Vector).
        Vector Database (VDB): Nơi lưu trữ các dãy số này (như ChromaDB, Pinecone, Weaviate, milvus). Đây giống như một "thư viện số" cho phép tìm kiếm theo ý nghĩa thay vì tìm theo từ khóa.
